```python
import pandas as pd

# Load the dataset
df = pd.read_csv("aqi.csv")

# Ask the user for the place
place = input("Enter the place: ").strip()

# Find the place
data = df[df["area"].str.lower() == place.lower()]

if data.empty:
    print("\nPlace not found in the dataset")

else:
    # Take the first record for that place
    row = data.iloc[0]

    # Get AQI
    aqi = row["aqi_value"]

    print("\n--- AQI SIMPLE REFLEX AGENT ---")

    print("Place:", row["area"])
    print("State:", row["state"])
    print("AQI:", aqi)

    # Simple if-else conditions
    if aqi <= 50:
        condition = "Good"
        status = "Air quality is good"

    elif aqi <= 100:
        condition = "Satisfactory"
        status = "Air quality is good"

    elif aqi <= 200:
        condition = "Moderate"
        status = "Air quality is not good"

    elif aqi <= 300:
        condition = "Poor"
        status = "Air quality is not good"

    elif aqi <= 400:
        condition = "Very Poor"
        status = "Air quality is not good"

    else:
        condition = "Severe"
        status = "Air quality is not good"

    print("Condition:", condition)
    print("Status:", status)
```
